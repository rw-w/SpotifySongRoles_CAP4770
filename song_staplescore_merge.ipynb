```python
# This script merges song features with staple scores and fills missing values. This can be used for further analysis.
# However, due to the number and sparsity of themes discovered, it is not used in the clustering analysis as it introduces way too many dimensions.
# Instead, the clustering analysis uses the song features without staple scores.
# Nonetheless, the staple scores are still useful for other analyses, such as individual song role classification within a theme.
import pandas as pd
import sqlite3

conn = sqlite3.connect('dbs/song_features.db')
song_features_df = pd.read_sql_query("SELECT * FROM song_features", conn)
conn.close()

staple_scores_df = pd.read_pickle('dbs/staple_scores.pkl')
staple_scores_df = staple_scores_df.reset_index().rename(columns={'index': 'track_uri'})

merged_df = pd.merge(
    song_features_df,
    staple_scores_df,
    how='left',
    on='track_uri'
)

theme_columns = staple_scores_df.columns.difference(['track_uri'])
merged_df[theme_columns] = merged_df[theme_columns].fillna(0)
```
